#Оформлення print()


```python
name = "Viktor"
language = "english"
print("Hi, I'm",name,"and I'm learning",language,"!")
print('Hi,',name,'!')
```

Це використання аргументів до функції print.

Рядок розбивається на кілька частин *"текст"* або *'текст'* , а комами позначають значення змінних.
Кожна частина з’єднується пробілом автоматично print.Тому потім можно побчити зяйви пропуски.

```python
name = "Viktor"
language = "english"
print(f"Hi, I'm {name} and I'm learning {language}!")
print(F'Hi,{name}!')
```

У цьому прикладі f-стрічка (f-string) використовується для форматованого виведення в Python.

Що таке f-стрічка:

Перед рядком ставиться префікс f або F: f"..." або f'...'.
Усередині такої рядки можна вставляти вирази в фігурних дужках { }.
Під час виконання вирази всередині дужок замінюються їхніми значеннями.

Переваги f-стрічок:

Легке вставлення змінних.
Можна також записувати вирази всередині дужок, наприклад: {2 + 3}, {len(language)}.

```python
print(f"Hi, 2+2={2+2}!")
a = "lenguage"
print(F'lenguage={len(a)} символів')
```

# Оформлення input() для символів, строк.



Вводимо значення через Enter. Стовпчиком.

```python
n = input("ввести ім'я:")
p = input("ввести прізвище:")
print(n,p)
```

Якщо потрібно вводити значення символів в один рядок через пропуск.


```python
n,p = input().split()
print(n,p)
```

розбиття рядка на дві змінні за допомогою методу split.

input() зчитує рядок з введення користувача.
.split() розбиває цей рядок на список підрядків за пробілом за замовчуванням.

Якщо користувач введе кілька слів, розбиті частини будуть у списку.
Якщо ви чітко очікуєте рівно дві частини, як у прикладі, Python розпакує список на дві змінні a і b.
Якщо введено більше або менше ніж дві частини, отримає помилку ValueError: not enough values to unpack (expected 2, got N) або too many values to unpack.



Якщо потрібно зчитати дві числа з одного рядка
12 34  


```python
a,b = map(int,input().split())
```

input() зчитує рядок, split() розбиває на частини за пробілом, map(int, ...) перетворює кожну частину на ціле число, а кортеж з двох значень розпаковується в a і b.



Зчитати дві числа з одного рядка: "12 34"

```python
a, b = map(int, input().split())
```

Пояснення: input() зчитує рядок, split() розбиває на частини за пробілом, map(int, ...) перетворює кожну частину на ціле число, а кортеж з двох значень розпаковується в a і b.

З використанням maxsplit на випадок зайвих слів Користувач може випадково ввести більше двох чисел,але потрібно взяти лише перші два python

```python
a, b = map(int, input().split(maxsplit=1))
```

Проте maxsplit в цій конфігурації застосовується так, щоб розбити максимум на два елементи:
перше число і решта в рядок. Ліпше використати явне розділення та перевірку кількості частин.

Якщо потрібно обробити лише два цілих числа, але можуть бути додаткові пробіли

```python
parts = input().split()
if len(parts) != 2:
  print("Будь ласка введіть exactly два цілі числа через пробіл.")
else: a, b = map(int, parts)
```
